# Common Regex Patterns in Python

## 1. `\d+`
**Meaning:** One or more digits.

**Examples:** `2026`, `123`, `98765`

## 2. `\s+`
**Meaning:** One or more whitespace characters, such as spaces, tabs, or newlines.

**Examples:** Extra spaces, tabs, multiple line breaks.

## 3. `https?://\S+`
**Meaning:** A simple URL pattern that matches URLs beginning with `http://` or `https://`.

**Examples:** `https://example.com`, `http://google.com`

## 4. `<.*?>`
**Meaning:** A basic HTML-tag pattern that matches text starting with `<` and ending at the nearest `>`.

**Examples:** `<p>`, `</p>`, `<h1>`

## 5. `[^a-zA-Z\s]`
**Meaning:** Matches any character other than English letters (`a-z`, `A-Z`) and whitespace.

**Examples:** `123`, `!`, `@`, `#`

**Note:** This pattern matches one character at a time. Use `+` (`[^a-zA-Z\s]+`) to match one or more consecutive characters that are not English letters or whitespace.

```python
# Raw text
text = """<p> WOW!!! I bought 2 phones in 2026.
Visit https://shop.com for more info.
This product is AMAZING!!! </p>"""
```

```python
# Convert text to lowercase
text = text.lower()
text
```

```python
import re
```

```python
# Remove URL
text = re.sub(f'https?://\S+', '', text)
text
```

```python
# Remove number
text = re.sub(r'\d+', "", text)
print(text)
```

```python
# Remove punctuation and special characters
text = re.sub(r'[^a-zA-Z\s]', "", text)
print(text)
```

```python
# Stop Word Removal

# Original Text
text = 'The product is very good and easy to use'
```

```python
# Stop words
stop_words = {'the', 'is', 'very', 'and', 'to'}
```

```python
# Cleaned Text
cleaned_text = []
```

```python
text = text.lower().split()
```

```python
for word in text:
  if word not in stop_words:
    cleaned_text.append(word)
```

```python
cleaned_text
```

# Stemming

```python
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
text = 'The product is very good and easy to use'
tokens = word_tokenize(text)
lemmatized_words = [lemmatizer.lemmatize(word) for word in tokens]

print(f"Original Text: {text}")
print(f"Lemmatized Words: {lemmatized_words}")
```

```python
from nltk.tokenize import word_tokenize
from nltk import pos_tag
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
sentence = "The children are running towards a better place."
tokens = word_tokenize(sentence)
tagged_tokens = pos_tag(tokens)

def get_wordnet_pos(tag):
    if tag.startswith('J'):
        return 'a'
    elif tag.startswith('V'):
        return 'v'
    elif tag.startswith('N'):
        return 'n'
    elif tag.startswith('R'):
        return 'r'
    else:
        return 'n'

lemmatized_sentence = []
for word, tag in tagged_tokens:
    if word.lower() == 'are' or word.lower() in ['is', 'am']:
        lemmatized_sentence.append(word)
    else:
        lemmatized_sentence.append(
            lemmatizer.lemmatize(word, get_wordnet_pos(tag)))
print("Original Sentence: ", sentence)
print("Lemmatized Sentence: ", ' '.join(lemmatized_sentence))
```

# Bag of Words

```python
# Exercise 1: Build the Vocabulary
```

```python
reviews = ['food was good',
           'delivery was late',
           'food was cold and late']
```

```python
vocab = []
```

```python
for review in reviews:
  print(review)
```

```python
for review in reviews:
  for word in review.lower().split():
    print(word)
```

```python
for review in reviews:
  for word in review.lower().split():
    if word not in vocab:
      vocab.append(word)
print(vocab)
```

```python
# Exercise 2: Build the BoW matrix
```

```python
bow = []
for review in reviews:
  tokens = review.lower().split()
  print(tokens)
```

```python
for review in reviews:
  tokens = review.lower().split()
  vector = []
  for word in vocab:
    vector.append(tokens.count(word))
    bow.append(vector)
  print(vector)
```
